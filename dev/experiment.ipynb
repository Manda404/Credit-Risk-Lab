```python
"""
==============================================================
📘 METRIQUES DE CLASSIFICATION POUR CATBOOST & XGBOOST
==============================================================

Les modèles de gradient boosting (CatBoost / XGBoost) permettent 
d’utiliser un large éventail de métriques pour monitorer 
l’apprentissage (`eval_metric`) et optimiser la performance 
sur les dataset équilibrés ou déséquilibrés.

Ci-dessous : les métriques les plus fiables, leurs définitions 
et quand les utiliser.

--------------------------------------------------------------
1️⃣ METRIQUES CLASSIQUES
--------------------------------------------------------------
✔ Accuracy ("Accuracy")
   - Score simple : proportion correcte de prédictions.
   - À éviter sur un dataset déséquilibré.
   (Ex : 90% négatifs → accuracy = 90% même sans apprendre.)

✔ Precision ("Precision")
   - Importance quand faux positifs = coûteux.
   - Exemple : accepter de bons clients uniquement.

✔ Recall ("Recall")
   - Importance quand faux négatifs = très graves.
   - Exemple : détecter les clients risqués → scoring crédit.

✔ F1 Score ("F1")
   - Compromis précision / recall.
   - Recommandé sur dataset déséquilibrés.

✔ Balanced Accuracy ("BalancedAccuracy")
   - Moyenne du recall par classe.
   - Indispensable si dataset déséquilibré.

--------------------------------------------------------------
2️⃣ METRIQUES ROBUSTES POUR DATASET DÉSÉQUILIBRÉ
--------------------------------------------------------------
✔ AUC ("AUC")
   - Standard industriel en scoring (banque, assurance).
   - Indépendant du seuil de décision.
   - Excellent pour datasets déséquilibrés.

✔ PRAUC ("PRAUC")
   - AUC de la courbe Precision-Recall.
   - Encore meilleur que AUC si classe minoritaire très faible.

✔ MCC ("MCC")
   - “Matthews Correlation Coefficient”.
   - La meilleure métrique globale sur dataset déséquilibré.
   - Symétrique : punit vraiment les erreurs.

✔ Kappa ("Kappa")
   - Mesure la qualité du modèle vs. hasard.
   - Utile si dataset très déséquilibré.

✔ Normalized Gini ("NormalizedGini") — CatBoost uniquement
   - Très utilisé en modélisation de risque crédit.
   - Equivaut à 2*AUC - 1.

--------------------------------------------------------------
3️⃣ METRIQUES LIÉES À LA PROBABILITÉ
--------------------------------------------------------------
✔ LogLoss ("Logloss") ou CrossEntropy
   - Mesure la qualité des probabilités prédites.
   - Indispensable pour calibration.

✔ Brier Score ("BrierScore")
   - Mesure la précision de la probabilité estimée.

--------------------------------------------------------------
4️⃣ QUELLE METRIC UTILISER SELON LE CONTEXTE ?
--------------------------------------------------------------
➡ Dataset équilibré :
   - Accuracy, F1, AUC

➡ Dataset modérément déséquilibré :
   - AUC, F1, BalancedAccuracy

➡ Dataset fortement déséquilibré :
   - MCC (recommandé)
   - PRAUC
   - Recall (si FN graves)
   - NormalizedGini (banque / assurance)

➡ Scoring crédit :
   - AUC
   - NormalizedGini
   - LogLoss

--------------------------------------------------------------
5️⃣ CONSEIL PRATIQUE
--------------------------------------------------------------
• CatBoost accepte n’importe quel nom de metric : "AUC", "F1", 
  "BalancedAccuracy", "Kappa", etc.

• XGBoost accepte les metrics suivantes : "logloss", "auc", 
  "aucpr", "error", "map", "ndcg", "merror", "mlogloss", etc.

• Avec les datasets déséquilibrés, évite Accuracy.
  Préfère : AUC, F1, MCC ou BalancedAccuracy.

--------------------------------------------------------------
6️⃣ RECOMMANDATION PAR DÉFAUT
--------------------------------------------------------------
Pour un dataset déséquilibré (comme le tien) :
    eval_metric = "AUC"
ou :
    eval_metric = "MCC"     (si FN & FP très importants)

==============================================================
"""
```
